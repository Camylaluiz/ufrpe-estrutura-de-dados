3. **Faça um programa que calcule e mostre a soma de dois números inteiros.**

```python
num1 = int(input("Digite o primeiro número inteiro: "))
num2 = int(input("Digite o segundo número inteiro: "))

soma = num1 + num2

print(f"A soma de {num1} e {num2} é: {soma}")
```

3_1. **Faça um programa que receba quatro números inteiros, calcule e mostre a soma destes
números;**

```python
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))
num4 = int(input("Digite o quarto número: "))

soma = num1 + num2 + num3 + num4

print(f"A soma dos quatro números é: {soma}")
```

4. **Faça um programa que receba três notas, calcule e mostre a média aritmética entre elas;**


```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = (nota1 + nota2 + nota3) / 3

print(f"A média aritmética das notas é: {media:.2f}")
```

5. **Faça um programa que receba O salário de um funcionário, calcule e mostre o novo
salário, sabendo-se que este sofreu um aumento de 25%;**


```python
salario_atual = float(input("Digite o salário atual do funcionário: R$ "))
novo_salario = salario_atual * 1.25
print(f"O novo salário com 25% de aumento é: R$ {novo_salario:.2f}")
```

6. **Faça um programa que calcule e mostre a área de um triangulo.
Sabe-se que: Área = (base * altura )/2;**


```python
base = float(input("Digite a base do triângulo: "))
altura = float(input("Digite a altura do triângulo: "))

area = (base * altura) / 2

print(f"A área do triângulo é: {area:.2f}")
```

7. **Faça um programa que calcule e mostre a área de um círculo.
Sabe-se que: Área = π r2**


```python
import math

raio = float(input("Digite o raio do círculo: "))
area = math.pi * (raio ** 2)
print(f"A área do círculo é: {area:.2f}")
```
